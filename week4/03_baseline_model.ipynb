```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
```

```python
df = pd.read_csv("../Week3/version2/cleaned_housing_data.csv")

print("=" * 60)
print("Data loaded successfully.")
print("=" * 60)

print(f"Dataset shape: {df.shape}")
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")

display(df.head())
```

```python
required_cols = ["Dataset", "ClosePrice"]

for col in required_cols:
    if col not in df.columns:
        raise ValueError(f"Required column '{col}' is missing.")

print("\nDataset split counts:")
print(df["Dataset"].value_counts(dropna=False))

print("\nClosePrice summary:")
display(df["ClosePrice"].describe())
```

```python
train_df = df[df["Dataset"] == "Train"].copy()
test_df = df[df["Dataset"] == "Test"].copy()

print("=" * 60)
print("Train/test split restored.")
print("=" * 60)

print(f"Train shape: {train_df.shape}")
print(f"Test shape:  {test_df.shape}")
```

```python
y_train = train_df["ClosePrice"].copy()
y_test = test_df["ClosePrice"].copy()

X_train = train_df.drop(
    columns=["ClosePrice", "Dataset"]
).copy()

X_test = test_df.drop(
    columns=["ClosePrice", "Dataset"]
).copy()
```

```python
print("=" * 60)
print("Features and target separated.")
print("=" * 60)

print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")

print(f"X_test shape:  {X_test.shape}")
print(f"y_test shape:  {y_test.shape}")
```

```python
print(
    "Train/test columns identical:",
    X_train.columns.equals(X_test.columns)
)
```

```python
non_numeric_cols = X_train.select_dtypes(
    exclude=["number", "bool"]
).columns.tolist()

print(f"Number of non-numeric feature columns: {len(non_numeric_cols)}")
print("Non-numeric columns:", non_numeric_cols)
```

```python
bool_cols = X_train.select_dtypes(include="bool").columns

X_train[bool_cols] = X_train[bool_cols].astype(int)
X_test[bool_cols] = X_test[bool_cols].astype(int)

print(f"Boolean columns converted to integers: {len(bool_cols)}")
```

```python
X_train = X_train.apply(pd.to_numeric, errors="coerce")
X_test = X_test.apply(pd.to_numeric, errors="coerce")

y_train = pd.to_numeric(y_train, errors="coerce")
y_test = pd.to_numeric(y_test, errors="coerce")
```

```python
print("=" * 60)
print("Missing value check")
print("=" * 60)

print(
    "Missing values in X_train:",
    X_train.isna().sum().sum()
)

print(
    "Missing values in X_test:",
    X_test.isna().sum().sum()
)

print(
    "Missing values in y_train:",
    y_train.isna().sum()
)

print(
    "Missing values in y_test:",
    y_test.isna().sum()
)
```

```python
train_inf_count = np.isinf(X_train.to_numpy()).sum()
test_inf_count = np.isinf(X_test.to_numpy()).sum()

print(f"Infinite values in X_train: {train_inf_count}")
print(f"Infinite values in X_test: {test_inf_count}")
```

```python
linear_model = LinearRegression()

linear_model.fit(X_train, y_train)
```

```python
train_pred = linear_model.predict(X_train)
test_pred = linear_model.predict(X_test)

print("Train prediction shape:", train_pred.shape)
print("Test prediction shape:", test_pred.shape)

assert len(train_pred) == len(y_train)
assert len(test_pred) == len(y_test)
```

```python
train_r2 = r2_score(y_train, train_pred)
test_r2 = r2_score(y_test, test_pred)

train_rmse = np.sqrt(
    mean_squared_error(y_train, train_pred)
)

test_rmse = np.sqrt(
    mean_squared_error(y_test, test_pred)
)

train_mae = mean_absolute_error(
    y_train,
    train_pred
)

test_mae = mean_absolute_error(
    y_test,
    test_pred
)
```

```python
print("=" * 60)
print("Linear Regression Baseline Results")
print("=" * 60)

print(f"Train R²   : {train_r2:.4f}")
print(f"Test R²    : {test_r2:.4f}")

print()

print(f"Train RMSE : ${train_rmse:,.2f}")
print(f"Test RMSE  : ${test_rmse:,.2f}")

print()

print(f"Train MAE  : ${train_mae:,.2f}")
print(f"Test MAE   : ${test_mae:,.2f}")
```

```python
comparison = pd.DataFrame({
    "Actual": y_train,
    "Predicted": train_pred
})

comparison["Error"] = comparison["Predicted"] - comparison["Actual"]
comparison["AbsoluteError"] = comparison["Error"].abs()

# Absolute percentage error (%)
comparison["PercentageError"] = (
    comparison["AbsoluteError"] / comparison["Actual"] * 100
)

comparison.sample(10)
```

```python
comparison.sort_values(
    "AbsoluteError",
    ascending=False
).head(20)
```

```python
print("Minimum train prediction:", train_pred.min())
print("Maximum train prediction:", train_pred.max())
print("Minimum test prediction:", test_pred.min())
print("Maximum test prediction:", test_pred.max())
```
